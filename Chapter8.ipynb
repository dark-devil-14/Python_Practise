<h3> </bold>**The self Parameter**</bold> </h3> 

    The self parameter is a reference to the current instance of the class.

    It is used to access properties and methods that belong to the class.

**Self refers to the instance of the class It  automatically passed with the function  clas from the object**

**Example 1**

```python
# Use self to access class properties:

class emp:
    lang = "Py"
    sal = "1,232,44"

    #create a func inside a class
    def getInfro(self):
        print(f"The language is {self.lang}, and the salary is {self.sal}")


harry = emp() 
print(harry.lang, harry.sal) 
harry.getInfro() #OR emp.getInfo(harry)
```

**Example 2**

```python
class person:
    # This defines a class called person (a blueprint)

    def name_l(self, name):
        # name_l is a method of the class
        # self refers to the object that is calling this method
        # name is the value passed from outside

        self.name = name
        # Store the value of name INSIDE the object
        # After this line, the object has a variable called name

    def print_name(self):
        # Another method of the class
        # self again refers to the current object

        print(self.name)
        # Print the name stored inside THIS object


rajuram = person()
rajuram.name_l("Rajuram")
# Create an object named rajuram from the person class

rajuram.print_name()
# Call print_name using the object rajuram
# self = rajuram
# ERROR happens here because rajuram.name was never set
```

<H3> The __init__() Method </h3>

    All classes have a built-in method called __init__(), which is always executed when the class is being initiated.

    The __init__() method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

```python
class emp:
    sal = 12,23,121
    lan = "py"

    def __init__(self): #dunder method which is automatically called
        print("i am creating an object")

    def get_info(self):
        print(f"This is a salary {self.sal} and language is {self.lan}", self.sal, self.lan)

    @staticmethod
    def greet():
        print("Good Morning")

harry = emp()
harry.name = "Harry"

print(harry.sal, harry.name, harry.lan)
harry.greet()
harry.get_info()
```
