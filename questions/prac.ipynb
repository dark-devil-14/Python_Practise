```python
#Example Determine if the year is the leap year using nested condititonal statements

year  = int(input("Enter the years : "))

if year%4 == 0:
    if year%100 ==0:
        if year%400 == 0:
            print("Given year is a leap year")
        else:
            print("This year is not a leap year")
    else:
        print("year is a leap year")
else:
    print("This year is not a leap year")
```

```python
# Convert temp

def convert_temp(temp, unit):
    """This Function convert temperature between farenhite and Celcius"""
    if unit== 'C':
        return (temp * 9/5) + 32 #Convert celcius to farenhite 
    elif unit == 'F':
        return (temp - 32)*5/9 #Convert farenchite to celcius 
    else:
        return None

print(convert_temp(25, 'C'))
print(convert_temp(77, 'F'))
    
```

```python
#Password Strength Chceker

print("Enter your passowrd")

def is_strong_password(password):
    """Checks if the passwoed is strong or not"""
    if len(password)<8: #Length of the password 
          return False
    if not any(char.isdigit() for char in password): #Numbers in password
          return False
    if not any(char.islower() for char in password): #lower string in password 
          return False
    if not any(char.isupper() for char in password): #upper string in password 
          return False
    if not any(char in '@#$%^&*!'for char in password): #Special string in password  
        return False 
    return True


is_strong_password("Rajuram9314@")
     

```

```python
# <!-- #Calculate the total cost of the itema in a shopping cart  -->

def calcualte_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost += item['Price']*item['quantity']
    return total_cost

cart = [
    { 'name' : 'Apple', 'Price': 6.34, 'quantity': 3 },
    {'name' : 'Orange', 'Price': 2.34, 'quantity': 34},
    {'name' : 'Bannana', 'Price': 3.34, 'quantity': 8}
]

calcualte_total_cost(cart)
```
