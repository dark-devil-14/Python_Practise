Practise Problem 1 

    create a class programmer for storing information of few programmers working at micrsoft



```python
class Programmer:
    company = "Microsoft"

    def __init__(self, name, salary, email, pincode):
        self.name = name
        self.salary = salary
        self.email = email
        self.pincode = pincode


programmers = [
    Programmer("Harry", "1,21,221", "harry@microsoft.com", "560026"),
    Programmer("Rajuram", "91,221", "rajuramd@microsoft.com", "562031"),
    Programmer("Manish", "1,54,231", "dmanish@microsoft.com", "52316"),
    Programmer("James", "87,221", "v.james@microsoft.com", "55921"),
    Programmer("Camila", "1,13,432", "gracy.camila@microsoft.com", "560026"),
    Programmer("Sam Altman", "1,82,541", "altman.sam@microsoft.com", "560068"),
]

print(f"{'Name':<15} {'Company':<12} {'Email':<30} {'Pincode':<8} {'Salary':>10}")
print("-" * 80)

for p in programmers:
    print(f"{p.name:<15} {p.company:<12} {p.email:<30} {p.pincode:<8} {p.salary:>10}")

```

Problem Statmeent 2 

    Create a class "calculator" capable of finding square cibe and square root of number

```python
class calculator:

    def __init__(self, n):
        self.n = n

    def square(self):
        print(f"The square is {self.n * self.n}")

    def square_root(self):
        print(f"The square root is {self.n ** 0.5}")

 
a = calculator(4)
a.square()
a.square_root()
```
