<h3> Inheritance and Advance OOPs Concept  </h3>

    Inheritance allows us to define a class that inherits all the methods and properties from another class.

    Parent class is the class being inherited from, also called base class.

    Child class is the class that inherits from another class, also called derived class.


```python
class emp:
    company = "ITC"
    name = "Rajuram"
    def show(self):
         print(f"The name is {self.name} and the Comapany is {self.company}")

# class programmer:
#     company = "ITC INFOTECH"
#     def show(self):
#          print(f"The name is {self.salary} and the salary is {self.salary}")
    
#     def showLanguage(self):
#         print(f"The name is {self.name} and he is good at {self.language}")


#Inherited class
class programmer(emp): #derived class
    company = "ITC INFOTECH"
    language = "Python"
    def showLanguage(self):
        print(f"The name is {self.name} and he is good at {self.language}")

a = emp()
b = programmer()


b.showLanguage()
print(a.company, b.company)

        
```

Multiple Inheritance 

```python
class company_data:
    company = "Microsoft"

    def printCompany(self):
        print(f"The name of the employer is: {self.company}")


class employee_data:
    emp_name = "Rajuram"
    emp_address = "Singasandra Bengaluru"
    emp_sal = "1,23,122"
    emp_email = "d.rajuram@msft.com"
    emp_doj = "21-3-2027"

    def printEmployee(self):
        print(
            f"Employee Name: {self.emp_name}\n"
            f"DOJ: {self.emp_doj}\n"
            f"Email: {self.emp_email}\n"
            f"Salary: {self.emp_sal}\n"
            f"Address: {self.emp_address}"
        )


class Onboarding_data(company_data, employee_data):

    user_laptop = "Micro-Binge-21"
    laptop_ip = "21.22.31.01"

    def onboarding(self):
        self.printCompany()
        print()
        print(f"{self.emp_name}!! Congratulations, welcome to {self.company}")
        print()
        self.printEmployee()
        print()
        print("Laptop Details:")
        print(self.user_laptop)
        print(self.laptop_ip)


Usr = Onboarding_data()
Usr.onboarding()
```
