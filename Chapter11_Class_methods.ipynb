```python
# @classmethod

class emp:
    a = 1

    @classmethod #use cls in func 
    def show(cls):
        print(f"The classs attribute value of a is {cls.a}")

e = emp()
e.a = 45 #instance attribute take prefrence but u want the class attribute only
e.show()
```
