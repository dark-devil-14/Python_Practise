```python
class Student:
    # Class attribute (should be a string)
    college_name = "IIT Guwahati"

    # Parameterized constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        print(f"Creating New Student: {self.name}")
        print(f"Marks: {self.marks}")

# Get number of students
case = int(input("Enter the no of students: "))

# Use a list or dictionary to store multiple student objects
students_list = []

for i in range(case):
    print(f"\n--- Entry for Student {i+1} ---")
    print(Student.college_name)
    name_input = input("Enter Name: ")
    marks_input = float(input("Enter Marks: "))
    
    # Create the object and add it to our list
    new_student = Student(name_input, marks_input)
    students_list.append(new_student)


for i in range(case):
    print("-------------------------\n",students_list[i].name, students_list[i].marks)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_average(self):
        print(f"Hi .. {self.name}, Your average score is {sum(self.marks)/3} ")


s1 = Student("Rajuram", [32,88,66])
s1.get_average()

s1.name = "Ironman"
s1.get_average()
```

```python
#Statics method changes the behavious of the fucntion where we no have to write self all time in function 

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @staticmethod
    def get_average():
        print("HEllod words ")


s1 = Student("Rajuram", [32,88,66])
s1.get_average()

```

```python
#Abstraction : hiding the implementation details of a class and only showing the esssential features of the user 

#Encapsulation : Wraping data and function into a single unit(object)

class car:
    def __init__(self): #Constructor
        self.esclator = False
        self.brk = False
        self.clutch = False

    def start(self):
        self.clutch = True
        self.esclator = True
        print("Car Started.....")


car1 = car()
car1.start()

#This is the example of the abstraction where all the details of related to car is hidden but user get to see only main code
```

```python
# To create a private attribute you use __ after the self. in attribute of class

class BankAccount:
    def __init__(self, account_no, password):
        # Private attributes
        self.__account_no = account_no
        self.__password = password
        self.balance = 0  # Public attribute

    # Public method to access data securely (Getter)
    def get_account_details(self, input_password):
        if input_password == self.__password:
            return f"Account: {self.__account_no}"
        else:
            return "Invalid Password!"

# Creating an instance
user = BankAccount("123456789", "SecurePass123")

# 1. Direct access fails
# print(user.__account_no)  # This would raise an AttributeError

# 2. Access through a method works
print(user.get_account_details("SecurePass123")) 

# 3. The "Secret" access (How Python hides it)
print(user._BankAccount__account_no) # This still works!
```

Private attributes & method are meant to be used only within the class and arenot acessible from outside the class
